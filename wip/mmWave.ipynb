```python
import serial
CLI_PORT = "COM6"
DATA_PORT = "COM5"

cli = serial.Serial(CLI_PORT, 115200, timeout=1)
data = serial.Serial(DATA_PORT, 921600, timeout=1)

import time
current_time = time.time()
while time.time() < current_time+.01:
    bytecount = data.inWaiting()
    s = data.read(bytecount)
    print(s)
cli.close()
data.close()
```

```python
import struct

def analyze_bytes(data):
    results = ["=== BYTE ANALYSIS REPORT ===", f"Total Length: {len(data)} bytes\n"]
    
    # 1. HEX DUMP & ASCII: Best for visual structure and finding text
    results.append("--- HEX DUMP & ASCII ---")
    results.append(f"{'Offset':<7} | {'Hexadecimal (4-byte groups)':<35} | {'ASCII'}")
    results.append("-" * 65)
    for i in range(0, len(data), 16):
        chunk = data[i:i+16]
        hex_parts = [' '.join(f'{b:02x}' for b in chunk[j:j+4]) for j in range(0, len(chunk), 4)]
        hex_str = '   '.join(hex_parts)
        ascii_str = ''.join(chr(b) if 32 <= b <= 126 else '.' for b in chunk)
        results.append(f"{i:04x}    | {hex_str:<35} | {ascii_str}")
    
    # 2. RAW INTEGERS: Helpful for identifying color values or signal levels
    results.append("\n--- RAW DECIMAL INTEGERS (List) ---")
    int_list = list(data)
    for i in range(0, len(int_list), 10):
        results.append(str(int_list[i:i+10]))
        
    # 3. STRUCT SPECULATION: Interprets chunks as 32-bit Integers and Floats
    results.append("\n--- SPECULATIVE DATA UNPACKING (4-byte chunks) ---")
    results.append(f"{'Offset':<7} | {'Int32':<12} | {'Float32'}")
    results.append("-" * 35)
    for i in range(0, len(data) - 3, 4):
        chunk = data[i:i+4]
        val_int = struct.unpack('<i', chunk)[0]
        val_float = struct.unpack('<f', chunk)[0]
        results.append(f"{i:04d}    | {val_int:<12} | {val_float:.6e}")
            
    return "\n".join(results)

# Your data
new_data = b'\x02\x01\x04\x03\x06\x05\x08\x07\x00\x00\x06\x03\x00\x03\x00\x00Ch\n\x00\x01\x00\x00\x00g\xe8QM\x05\x00\x00\x00\x05\x00\x00\x00\x00\x00\x00\x00\x01\x00\x00\x00P\x00\x00\x00\xcaH\x9c\xbc\x8e\x83\xce>/m\xea>\x00\x00\x00\x00\xc1\x84\xce\xbeM\xef\xba?E\xfb\r\xbf\x00\x00\x00\x00\x8fr\xa7\xbe\xdb\xcb\xd5?\xfcZ\xc3\xbe\x00\x00\x00\x00f\xd2k\xbe\x05\x80\x97>\x86\xae\xd9>\xf07w\xbef\xd2k\xbe\xf1\x8d\x89>v\xc0\xe2>\xf07w\xbe'

print(analyze_bytes(new_data))
```
